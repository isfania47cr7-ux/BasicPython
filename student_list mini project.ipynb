```python
students=[
[101,"Bob","Python",15000,"Active"],
[102,"Alice","Java",12000,"Active"],
[103,"David","Full Stack",20000,"Inactive"],
[104,"Charlie","Python",15000,"Active"]
]
while True:
    print("= "*8+"STUDENT COURSE MANAGEMENT"+"= "*8)
    print("1.Add New Student")
    print("2.Insert Student at Specific Position")
    print("3.Import students from Another Branch")
    print("4.View all Students")
    print("5.Search Student")
    print("6.Update Student Course")
    print("7.Remove Student by Name")
    print("8.Remove last enrolled student")
    print("9.Backup Student Records")
    print("10.Count Python Students")
    print("11.Sort Students by Name")
    print("12.Reverse Student List")
    print("13.Find student Position")
    print("14.Display first Five students")
    print("15.Display Last Three Students")
    print("16.Delete Student Using Index")
    print("17.Check Student Using Index")
    print("18.Clear All Student Records")
    print("19.Display Total Number of Students")
    print("20.Exit")

    choice=int(input("Enter your choice:"))
    print("Enter the Choice",choice)
    if choice == 1:
        print("Add New Student:")
        student_id=int(input("Enter Student id"))
        student_name=input("Student Name:")
        student_course=input("Enter student course")
        course_fee=int(input("Enter course fee:"))
        course_status=input("Enter Student Status")
        course_list=[student_id,student_name,student_course,course_fee,course_status]
        students.append(course_list)
        print("Student Inserted Successfully")
        for student in students:
           print(student,end=" \n ")
        print()
        
    elif choice ==2:
        position=int(input("Enter the Position:"))
        print("Insert Student at Required Position")
        student_id=int(input("Enter Student id"))
        student_name=input("Student Name:")
        student_course=input("Enter student course")
        course_fee=int(input("Enter course fee:"))
        course_status=input("Enter Student Status")
        course_list=[student_id,student_name,student_course,course_fee,course_status]
        students.insert(position,course_list)
        print("Student Inserted Successfully")
        for student in students:    
            print(student,end=" \n ")
        print()
    elif choice ==3:
        branch_students=[[105,"Eva","Python",15000,"Active"],
        [106,"Frank","Java",1200,"Active"]]
        students.extend(branch_students)
        for student in students:
            print(student,end=" \n ")
        print()
    elif choice ==4 :
        if len(students) != 0:
            print("-"*50)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("ID","NAME","COURSE","FEE","STATUS"))
            print("-"*50)
            for student in students:
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                    student[0],
                    student[1],
                    student[2],
                    student[3],
                    student[4]))
        else:
            print("Students Not Registered")

    elif choice== 5:
        search_name=input("Search Student by name:").lower().strip()
        print("Search Student By Name:",search_name)
        print("= "*10+"Student Details"+"= "*10)
        found = False #flag used
        for student in students:
            if student[1].lower().strip()==search_name:
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("ID","NAME","COURSE","FEE","STATUS"))
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                    student[0],
                    student[1],
                    student[2],
                    student[3],
                    student[4]))
                print("-"*50)
                found=True
        if not found:
            print("Student Name Not Found")
    elif choice == 6:
        search_name=input("Search Student by name:").lower().strip()
        print("Enter the student Name:",search_name)
        found=False
        for student in students:
            if student[1].lower().strip()==search_name:
                new_course=input("enter the new course:")
                print("enter The New Course:",new_course)
                student[2]=new_course
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("ID","NAME","COURSE","FEE","STATUS"))
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                student[0],
                student[1],
                student[2],
                student[3],
                student[4]))
                print("-"*50)
                found=True
        if not found:
            print("Student not found")
    elif choice == 7:
        search_name=input("Enter student name to remove").lower().strip()
        print("Enter Student Name To Remove:",search_name)
        found=False
        for student in students:
            if student[1].lower().strip() == search_name:
                students.remove(student)
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
                print("-"*50)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                student[0],
                student[1],
                student[2],
                student[3],
                student[4]
                ))
                print("-"*50)
                found=True
        
        if not found:
            print("student not found")
        

    elif choice == 8 :
        if len(students) != 0:
            last_student=students.pop()
            print("-"*50)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
            print("-"*50)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
            last_student[0],
            last_student[1],
            last_student[2],
            last_student[3],
            last_student[4]
            ))
            print("-"*50)
        else:
            print("student not found")
    elif choice == 9:
        backup_students=students.copy()
        backup_student=len(backup_students)
        if backup_student != 0:
            print("Backed up records successfully")
            print("Total Records Backed Up:",backup_student)
        else:
            print("No Records Found")
    elif choice == 10:
        count=0
        course_name=input("Enter Course Name To Search").lower().strip()
        print("Enter Course Name To Search:",course_name)
        
        print("-"*50)
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
        print("-"*50)
        found=False
        for student in students:
            if student[2].lower().strip() == course_name:
                count+=1
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                student[0],
                student[1],
                student[2],
                student[3],
                student[4]
                ))
                print("Total Python students:",count)
                found=True
            if not found:
                print("Course Not Found")
    elif choice == 11:
        print("-"*50)
        students.sort(key=lambda student:student[1].lower())
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))

        for student in students:
            
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
            student[0],
            student[1],
            student[2],
            student[3],
            student[4]
            ))
            print("-"*50)
    elif choice == 12:
        print("-"*50)
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
        print("="*50)
        for student in students[::-1]:
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
            student[0],
            student[1],
            student[2],
            student[3],
            student[4]
            ))
            print("-"*50)

    elif choice == 13:
        search_name=input("Enter student name to search").lower().strip()
        print("Enter Student Name To search:",search_name)
        print("-"*50)
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
        print("-"*50)
        found=False
        for student in students:
            if student[1].lower().strip() == search_name:
                position = students.index(student)
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                student[0],
                student[1],
                student[2],
                student[3],
                student[4]
                ))
                found=True
        if not found:
            print("Student Not Found")
        
        print("Student found at position:",position)
    elif choice == 14:
        print("-"*50)
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
        print("-"*50)
        for student in students[0:2]:
            branch_students=[[105,"Eva","Python",15000,"Active"],
            [106,"Frank","Java",1200,"Active"]]
            students.extend(branch_students)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
            student[0],
            student[1],
            student[2],
            student[3],
            student[4]
            ))
        print("-"*50)
    elif choice == 15:
        for student in students[-3:]:
            branch_students=[[105,"Eva","Python",15000,"Active"],
            [106,"Frank","Java",1200,"Active"]]
            students.extend(branch_students)
            print("-"*50)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
            print("-"*50)
            print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
            student[0],
            student[1],
            student[2],
            student[3],
            student[4]
            ))
            print("-"*50)
            students.reverse()
    elif choice == 16:
        index=int(input("Enter The Index:"))
        print("Enter The index:",index)
        if index >=0 and index < len(students):
                del students[index]
        else:
            print("invalid index")
    elif choice == 17:
        search_name=input("Enter student name to search").lower().strip()
        print("Enter Student Name To search:",search_name)
        print("-"*50)
        print("{:<5}{:<10}{:<18}{:<8}{:<7}".format("Id","NAME","COURSE","FEE","STATUS"))
        print("-"*50)
        found=False
        for student in students:
            if student[1].lower().strip() == search_name:
                print("student Available")
                print("{:<5}{:<10}{:<18}{:<8}{:<7}".format(
                student[0],
                student[1],
                student[2],
                student[3],
                student[4]
                ))
                found=True
        if not found:
            print("Student Not Available")

    elif choice == 18:
        students.clear()
        print("All student Records cleared Successfully")  
    elif choice == 19:
        total_students=len(students)
        print(total_students)
    elif choice == 20:
        print("Thank You")
        print("Student Course Management System Closed successfully")
    
        break
        
```
