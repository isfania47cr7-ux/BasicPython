```python
employees=int(input("Employee:"))
total_salary=[]

for employee in range(1,employees):
    salary=int(input("Salary:"))
    print(f"Employee {employee} salary:{salary}")
    total_salary.append(salary)

print("Total Salary:",sum(total_salary))
print("Highest Salary:",max(total_salary))


             
```

```python
products=int(input("Products:"))
total_bill=0
for product in range(1,products):
    price=int(input("Price:"))
    print(f"Price {product}:{price}")
    total_bill=total_bill+price
print("Total Bill = ",total_bill)

```

```python
players=int(input("Players:"))
total_score=0
for player in range(1,players+1):
    runs=int(input("Runs:"))
    print(f"Player {player} : {runs}")
    total_score += runs
print("Total Score : ",total_score)

```

```python
students=int(input("Number of students :"))
total_mark=0
for student in range(1,students+1):
    marks=int(input("Marks:"))
    print(f"Enter Mark {student} : {marks}")
    total_mark+=marks
    #total_mark.append(marks)
print("Total Marks:",total_mark)
print("Average Mark:",(total_mark)/students)
```
