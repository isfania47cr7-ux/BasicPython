```python
stu_name=input("Enter Student Name:") 
stu_id=input("Student Id")
Test_score=int(input("Enter Test Score:"))

print("Enter Student Name  : ",stu_name)

print("Enter Student Id    : ",stu_id)

print("Enter Student Score : ",Test_score)
print()
print("= "*5 +"REPORT" + "= "*5)
print("Name  : ",stu_name)
print("ID    : ",stu_id)
print("Score : ",Test_score)

if (Test_score >= 95 and Test_score <= 100):
    print("GRADE :  A")
elif(Test_score <= 94 and Test_score >= 90):
    print("GRADE :  B")
elif(Test_score <= 89 and Test_score >= 85):
    print("GRADE :  C")
elif(Test_score <= 84 and Test_score >= 80):
    print("GRADE :  D")
elif(Test_score <= 79 and Test_score >= 70):
    print("GRADE :  E")
else:
    print("GRADE :  F")
```
