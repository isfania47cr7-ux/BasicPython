```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
student1=Student("Vijin",46)#student1.name="vijin"#instant variable
student2=Student("akhil",23)#student1.age=age
print(student1.name)
print(student2.name)
```

```python
class Student:
    college_name="ABC College"#class variable
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print("Name:",self.name)
        print("Age:",self.age)#instant method
student1=Student("Vijin",46)#student1.name="vijin"#instant variable
student2=Student("akhil",23)#student1.age=age
student1.display()
print(student1.name)
print(student2.name)
print(Student.college_name)
```
