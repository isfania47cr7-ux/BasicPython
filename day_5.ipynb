```python
rows=int(input("Rows:"))
seats=int(input("Seats:"))
total_seats=rows*seats
booked_seats=[]
count=0
while True:
    print("="*5 + "Seating  Layout  System" + "="*5)
    print("1.View Seats")
    print("2.Book Tickets")
    print("3.View Booked Tickets")
    print("4.View available seat Counts")
    print("5.Exit")
    choice=int(input("Enter Your choice:"))
    print("Enter your Choice:",choice)
    print()
    if choice == 1:
        for row in range(rows):
            row_seat=chr(65+row)
            for seat in range(1,seats+1):
                current_seat=f"{row_seat}{seat}"
                if current_seat in booked_seats:
                    print("X",end=" ")
                else:
                    print(current_seat,end=" ")
            print()
    if choice == 2:
        seat_number=input("Enter the Seat_number (Example:A1) :").upper()
        print("Enter Seat Number:",seat_number)
        valid_seat=False
        for row in range(rows):
                    row_seat=chr(65+row)
                    for seat in range(1,seats+1):
                        current_seat=f"{row_seat}{seat}"
                        if current_seat==seat_number:
                            valid_seat=True

        if not valid_seat:
            print("Invalid seat Number")
        elif seat_number in booked_seats:
            print("Seat Already Booked")
        else:
            booked_seats.append(seat_number)
            count+=1
            print("Ticket booked succesfully")
            print("="*5 + "Updated Theatre Layout" + "="*5 )
            for row in range(rows):
                row_seat=chr(65+row)
                for seat in range(1,seats+1):
                    current_seat=f"{row_seat}{seat}"
                    if current_seat in booked_seats:
                        print("X",end=" ")
                    else:
                        print(current_seat,end=" ")
                print()

 
    if choice ==3:
        break
print("Total Seats:",total_seats)
print("booked Seats:",count)
print("Available Seats:",(total_seats-count))

```
