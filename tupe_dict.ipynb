```python
total_products=int(input("Enter the Number of products"))
grand_total=0
products_list=[]
print("Enter Number of Products:",total_products)
print("="*40)
for pro_num in range(1,total_products+1):
    print("enter details of product:",pro_num)
    pro_id=int(input("product id"))
    pro_name=(input("Pro name"))
    price=int(input("Price:"))
    quantity=int(input("Quantity"))
    products_list.append((pro_id,pro_name,price,quantity))
    print(" product ID :",pro_id)
    print("Product Name:",pro_name)
    print("Price       :",price)
    print("Quantity    :",quantity)
    grand_total+=price*quantity

    
print("{:<5}{:<10}{:<15}{:<5}{:<10}".format("Id","NAME","PRICE","QTY","TOTAL"))
print("="*40)
for product in products_list:
    total_price=product[2]*product[3]
    print("{:<5}{:<10}{:<15}{:<5}{:<10}".format(
    product[0],
    product[1],
    product[2],
    product[3],
    total_price
    ))
print("="*40)
print("Grand Total:",grand_total)
```

```python
student={
    "name":"Isfania",
    "Age":21,
    "course":"Data science"
}
student["dept"]="CSE"
print(student)
print(student.get("dept"))
```

```python
student={
    "name":"Isfania",
    "Age":21,
}
student.update(
    {"course":"Datascience",
    "dept":"CSE"}
)
print(student)
```

```python
student={
    "name":"Isfania",
    "Age":21,
}
student.update(
    {"course":"Datascience",
    "dept":"CSE"}
)
student.pop("course")
student.popitem()
del student["name"]
print(student) #key value key_index
```
